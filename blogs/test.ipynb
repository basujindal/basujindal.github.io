```python
## find eigen values of image

import numpy as np
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('/home/thunder/my_github_clones/hugo-parent/content/blogs/images/hello_world.jpeg', 0)

# find eigen values using numpy
eigen_values = np.linalg.eigvals(img)


print(eigen_values.shape)

plt.imshow(img, cmap='gray')
```

```python
# print number of real eigen values
print(np.sum(np.isreal(eigen_values)))

eigen_values, eigen_vectors = np.linalg.eig(img)

# find index of real eigen values

real_eig = np.isreal(eigen_values)

### zero out all imaginary eigen values, keeping same shape as original eigen values
# eigen_values = eigen_values*real_eig

img_evd = np.dot(eigen_vectors, np.dot(np.diag(eigen_values), np.linalg.inv(eigen_vectors)))

img_evd = np.real(img_evd)

img_evd = (img_evd - np.min(img_evd))/(np.max(img_evd) - np.min(img_evd))

img_evd = np.uint8(img_evd*255)

plt.imshow(img_evd, cmap='gray')
plt.show()
```
